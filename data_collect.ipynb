```python
import cv2
import numpy as np

classifier = cv2.CascadeClassifier(
    r"C:\Users\patra\OneDrive\Desktop\FACE DETECTION\haarcascade_frontalface_default (1).xml"
)

# Open laptop camera
camera = cv2.VideoCapture(0)

data = []

while len(data) < 100:

    ret, frame = camera.read()

    if not ret:
        print("Could not access the camera")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    face_points = classifier.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5
    )

    if len(face_points) > 0:

        # Use the largest detected face
        x, y, w, h = max(
            face_points,
            key=lambda face: face[2] * face[3]
        )

        face_frame = frame[y:y+h, x:x+w]

        cv2.imshow("Only Face", face_frame)

        # Store face image
        data.append(face_frame)

        print(len(data), "/100")

        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (255, 0, 0),
            2
        )

    cv2.putText(
        frame,
        str(len(data)) + "/100",
        (50, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 0, 255),
        2
    )

    cv2.imshow("Laptop Camera", frame)

    # Press Q to stop
    if cv2.waitKey(30) & 0xFF == ord("q"):
        break

camera.release()
cv2.destroyAllWindows()


# Save collected images
if len(data) == 100:

    name = input("Enter Face holder name: ")

    for i in range(100):
        cv2.imwrite(
            "IMAGES/" + name + "_" + str(i) + ".jpg",
            data[i]
        )

    print("Done! Data saved successfully.")

else:
    print("Need more data")
```
